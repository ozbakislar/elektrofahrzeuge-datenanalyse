# 1. About the Dataset



This dataset contains technical and commercial performance data for electric vehicle (EV) models produced between 2020 and 2026. The dataset includes a total of 2000 observations and 24 columns, where each row represents a specific brand's model for a given year and variant (base, long range, premium, performance, standard).

**The brands included in the dataset are as follows:**

- **Germany:** Audi, BMW, Mercedes, Porsche, Volkswagen
- **USA:** Fisker, Ford, GM/Chevrolet, Lucid, Rivian, Tesla
- **Japan:** Honda, Toyota
- **South Korea:** Hyundai, Kia
- **China:** BYD, NIO, Xiaomi
- **Sweden:** Polestar, Volvo

## 5.3. Analysis of Relationships Between Variables

```python
corr = df.select_dtypes(include="number").corr()

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Matrix of Numerical Variables")

plt.show()
```

### 5.3.1. Range ↔ Battery Capacity

```python
sns.regplot(data=df, x="battery_capacity_kwh", y="range_miles", scatter_kws={"alpha": 0.5}, line_kws={"color": "red"})

plt.title("Relationship Between Battery Capacity and Range")
plt.xlabel("Battery Capacity (kWh)")
plt.ylabel("Range (Miles)")

plt.show()
```

```python
df["battery_capacity_kwh"].corr(df["range_miles"])
```

<div style='background-color: #e3f2fd; border-left: 5px solid #2196F3; padding: 15px; margin: 10px 0;'>
  <h3 style='margin-top: 0; color: #1565c0;'>🎯 Analysis</h3>
  <ul>
    <li>The scatter plot shows a very strong linear relationship between battery capacity and vehicle range. The dense clustering of points around the regression line clearly supports this relationship.</li>
    <li>The calculated Pearson correlation coefficient is <strong>r = 0.978</strong>, indicating a <strong>very strong positive</strong> relationship between the two variables.</li>
    <li>As battery capacity increases, vehicle range also increases significantly. This is consistent with the technical expectation that larger battery packs provide longer driving distances in electric vehicles.</li>
    <li>This analysis shows that battery capacity is one of the most important variables explaining vehicle range. It is also one of the strongest relationships observed in the heatmap analysis.</li>
  </ul>
</div>

# 6. Business Insights

As a result of this exploratory data analysis, the following key business insights were obtained on the electric vehicle dataset:

- There is a **very strong positive relationship (r = 0.978)** between **battery capacity** and **vehicle range**. This result shows that battery capacity is one of the most important technical factors determining vehicle range.

- **Engine power** has a **more pronounced effect on vehicle price than battery capacity**. While there is a **strong relationship between vehicle price and engine power (r = 0.765)**, the **relationship between battery capacity and price remains quite weak (r = 0.164)**.

- **Vehicles with longer range** tend to receive **higher customer ratings from users (r = 0.668)**. This shows that driving range is one of the important features influencing customer satisfaction.

- **Lucid**, **Porsche**, and **Mercedes** stand out as the brands with the highest average price levels. In contrast, **Volkswagen**, **Tesla**, and **BYD** are among the brands with the highest average sales volume. This finding shows that **a high price does not always mean high sales**.

- Although the **luxury** segment has the highest average vehicle price, the **budget** segment shows the highest average sales value. This shows that price level alone is not sufficient to explain sales performance.

- The numerical variable analyses show that extreme values such as high price, high engine power, and high customer rating in the dataset largely stem from **premium and high-performance vehicles**. Therefore, these observations were not treated as data errors.

- Overall, technical specifications, brand positioning, and market segment jointly influence vehicle price, sales performance, and customer satisfaction. These factors need to be considered together when evaluating the electric vehicle market.
