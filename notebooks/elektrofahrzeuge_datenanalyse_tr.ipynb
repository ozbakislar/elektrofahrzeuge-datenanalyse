# 1. Veri Seti Hakkında

**Veri setinde yer alan markalar şunlardır:**

- **Almanya:** Audi, BMW, Mercedes, Porsche, Volkswagen
- **ABD:** Fisker, Ford, GM/Chevrolet, Lucid, Rivian, Tesla
- **Japonya:** Honda, Toyota
- **Güney Kore:** Hyundai, Kia
- **Çin:** BYD, NIO, Xiaomi
- **İsveç:** Polestar, Volvo

### 5.1.4. Horsepower

```python
sns.histplot(data=df, x="horsepower", bins=20)

plt.axvline(df["horsepower"].mean(), color="red", linestyle="--", label="Ortalama")
plt.axvline(df["horsepower"].median(), color="green", linestyle="-", label="Medyan")

plt.title("Beygir Gücünün Dağılımı")
plt.xlabel("Beygir Gücü (HP)")
plt.ylabel("Frekans")

plt.legend()

plt.show()
```
