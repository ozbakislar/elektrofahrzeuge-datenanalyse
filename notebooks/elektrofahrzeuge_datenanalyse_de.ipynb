# 1. Überblick über den Datensatz



Dieser Datensatz enthält technische und kommerzielle Leistungsdaten von Elektrofahrzeug-Modellen (EV), die zwischen 2020 und 2026 produziert wurden. **Die im Datensatz enthaltenen Marken sind wie folgt:**

- **Deutschland:** Audi, BMW, Mercedes, Porsche, Volkswagen
- **USA:** Fisker, Ford, GM/Chevrolet, Lucid, Rivian, Tesla
- **Japan:** Honda, Toyota
- **Südkorea:** Hyundai, Kia
- **China:** BYD, NIO, Xiaomi
- **Schweden:** Polestar, Volvo

## 5.3. Analyse der Zusammenhänge zwischen den Variablen

```python
corr = df.select_dtypes(include="number").corr()

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Korrelationsmatrix der numerischen Variablen")

plt.show()
```

### 5.3.1. Range ↔ Battery Capacity

```python
sns.regplot(data=df, x="battery_capacity_kwh", y="range_miles", scatter_kws={"alpha": 0.5}, line_kws={"color": "red"})

plt.title("Zusammenhang zwischen Batteriekapazität und Reichweite")
plt.xlabel("Batteriekapazität (kWh)")
plt.ylabel("Reichweite (Meilen)")

plt.show()
```

```python
df["battery_capacity_kwh"].corr(df["range_miles"])
```

<div style='background-color: #e3f2fd; border-left: 5px solid #2196F3; padding: 15px; margin: 10px 0;'>
  <h3 style='margin-top: 0; color: #1565c0;'>🎯 Analyse</h3>
  <ul>
    <li>Das Streudiagramm zeigt einen sehr starken linearen Zusammenhang zwischen Batteriekapazität und Fahrzeugreichweite. Die dichte Häufung der Punkte um die Regressionsgerade untermauert diesen Zusammenhang deutlich.</li>

# 6. Geschäftliche Erkenntnisse

Als Ergebnis dieser explorativen Datenanalyse wurden die folgenden zentralen geschäftlichen Erkenntnisse zum Elektrofahrzeug-Datensatz gewonnen:

- Zwischen **Batteriekapazität** und **Fahrzeugreichweite** besteht ein **sehr starker positiver Zusammenhang (r = 0,978)**. Dieses Ergebnis zeigt, dass die Batteriekapazität einer der wichtigsten technischen Faktoren ist, die die Fahrzeugreichweite bestimmen.

- Die **Motorleistung** hat einen **stärkeren Einfluss auf den Fahrzeugpreis als die Batteriekapazität**. Während zwischen **Fahrzeugpreis und Motorleistung ein starker Zusammenhang (r = 0,765)** besteht, bleibt der **Zusammenhang zwischen Batteriekapazität und Preis recht schwach (r = 0,164)**.

- **Fahrzeuge mit größerer Reichweite** erhalten von den Nutzern tendenziell **höhere Kundenbewertungen (r = 0,668)**. Dies zeigt, dass die Reichweite eines der wichtigen Merkmale ist, die die Kundenzufriedenheit beeinflussen.

- **Lucid**, **Porsche** und **Mercedes** stechen als die Marken mit dem höchsten durchschnittlichen Preisniveau hervor. Demgegenüber gehören **Volkswagen**, **Tesla** und **BYD** zu den Marken mit dem höchsten durchschnittlichen Absatzvolumen. Dieser Befund zeigt, dass ein **hoher Preis nicht immer hohe Verkaufszahlen bedeutet**.

- Obwohl das Segment **luxury** den höchsten durchschnittlichen Fahrzeugpreis aufweist, zeigt das Segment **budget** den höchsten durchschnittlichen Absatzwert. Dies verdeutlicht, dass das Preisniveau allein nicht ausreicht, um die Absatzleistung zu erklären.

- Die Analysen der numerischen Variablen zeigen, dass extreme Werte wie hoher Preis, hohe Motorleistung und hohe Kundenbewertung im Datensatz größtenteils auf **Premium- und Hochleistungsfahrzeuge** zurückzuführen sind. Diese Beobachtungen wurden daher nicht als Datenfehler gewertet.

- Insgesamt betrachtet beeinflussen technische Spezifikationen, Markenpositionierung und Marktsegment gemeinsam den Fahrzeugpreis, die Absatzleistung und die Kundenzufriedenheit. Für die Bewertung des Elektrofahrzeugmarktes müssen diese Faktoren gemeinsam betrachtet werden.
